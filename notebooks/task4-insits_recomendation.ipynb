```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("data/processed/task2_final_output.csv")

plt.figure(figsize=(10,6))

sns.countplot(data=df, x="bank", hue="sentiment_label")

plt.title("Sentiment Distribution by Bank")
plt.xticks(rotation=45)
plt.show()
```

```python
sentiment_counts = df.groupby(["bank", "sentiment_label"]).size().unstack()

sentiment_counts.plot(kind="bar", stacked=True, figsize=(10,6))

plt.title("Sentiment Distribution by Bank")
plt.ylabel("Number of Reviews")
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(data=df, x="bank", y="rating")

plt.title("Rating Distribution by Bank")
plt.xticks(rotation=45)
plt.show()
```

```python
top_themes = df["identified_theme"].value_counts().head(10)

plt.figure(figsize=(8,5))
sns.barplot(x=top_themes.values, y=top_themes.index)

plt.title("Top Identified Themes")
plt.xlabel("Count")
plt.show()
```

```python
df["review_date"] = pd.to_datetime(df["review_date"])

trend = df.groupby(["review_date", "bank"])["sentiment_score"].mean().reset_index()

plt.figure(figsize=(10,6))
sns.lineplot(data=trend, x="review_date", y="sentiment_score", hue="bank")

plt.title("Sentiment Trend Over Time")
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.countplot(data=df, y="identified_theme", order=df["identified_theme"].value_counts().index)

plt.title("Theme Frequency")
plt.show()
```
